```python
# Создаем пустой список
squares = []

# Используем цикл for для перебора чисел от 1 до 10 включительно
# range(1, 11) генерирует числа от 1 до 10
for i in range(1, 11):
    # Вычисляем квадрат числа (i ** 2) и добавляем его в список
    squares.append(i ** 2)

# Выводим полученный список на экран
print(squares)
```

Используя цикл, создан список, содержащий квадраты всех целых чисел от 1 до 10, и выведит его на экран

временная сложность для этого кода O(1) (Константная). Цикл выполнится ровно 10 раз, независимо от любых внешних условий
пространственная сложность составляет О(N) (Линейная), так как объем потребляемой памяти растет прямо пропорционально количеству элементов
N

```python
squares = [i ** 2 for i in range(1, 11)]
print(squares)
```

```python
# 1. Создаем кортеж из 6 элементов разных типов данных
my_tuple = (
    42,                # int (целое число)
    3.14,              # float (число с плавающей точкой)
    "Привет, мир!",    # str (строка)
    [1, 2, 3],         # list (список)
    {"ключ": "значение"}, # dict (словарь)
    None               # NoneType (отсутствие значения)
)

# 2. Выводим только числовые элементы (int или float)
print("Числовые элементы кортежа:")
for item in my_tuple:
    # Проверяем, является ли тип элемента строго int или float
    if type(item) in (int, float):
        print(item)
```

этот пример создала для понимания сложности
